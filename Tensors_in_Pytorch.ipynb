```python
# Create a device variable
device=torch.device('cuda')
```

```python
# Create a tensor inside GPU
z=torch.randint(0,10, (3,2), device=device)
z
```

```python
# move existing tensor to gpu from cpu
z1=torch.randint(0,10, (3,2))
z1
c=z1.to(device=device)
c
```

```python
# Comparison of matrix maltiplication (cpu/GPU)
import time

# Define the matrix size
size=1000

# Creates matrix on cpu
matrix_cpu1=torch.randn(size, size)
matrix_cpu2=torch.randn(size, size)

# Creates matrix on gpu
matrix_gpu1=matrix_cpu1.to(device)
matrix_gpu2=matrix_cpu2.to(device)

# Measure the time on cpu
start_time_cpu=time.time()
result_cpu=torch.matmul(matrix_cpu1, matrix_cpu2)  # Matrix multiplication on cpu
cpu_timw=time.time()-start_time_cpu
print(f"Time on cpu: {cpu_timw:4f} seconds")

# Move matrix to gpu
matrix_gpu1=matrix_gpu1.to(device)
matrix_gpu2=matrix_gpu2.to(device)

# Measure the time on gpu
start_time_gpu=time.time()
result_gpu=torch.matmul(matrix_gpu1, matrix_gpu2)  # matrix multiplication on gpu
gpu_time=time.time()-start_time_gpu

print(f"Time on gpu: {gpu_time:4f} seconds")
```
