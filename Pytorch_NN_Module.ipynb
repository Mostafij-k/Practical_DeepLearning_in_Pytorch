```python
                    # ---------------torch.nn module--------#
                     
# It is a core module of Pytorch which contain various function and class to build Neural Network.
    
# Key components of torch.nn module

# 1. Modules (Layers)
    # nn Module: The base class for all neural network modules.
    # Common layers: Includes layers like nn.Linear(Fully Connected Layer), nn.Conv2D (Convolution Layer), nn.LSTM (Recurrent Layer)

# 2. Activation Function:
    #  Function Like nn.Relu() , nn.Sigmoid(), nn.Softmax(), nn.Tanh() introduce nonlinearities to the model.
    # Allow it to learn complex pattern of Data

# 3. Loss Function
    # Provides loss function such as nn.BinaryCrossEntrophy, nn.MSEloss, nn.NLLLoss to calculate the difference between predited value and actual value.

# 4. Container Modules:
    # nn.Sequential(): A sequential Container to stack layers in order. 

# 5. Regularization and Dropout:
    # Layers like nn.Dropout() and nn.BatchNorm2D help to prevent overfitting and improve the model performances.
    #  

```

```python
            # -----------torch.optim Module---------#
             
# It is a module in Pytorch that provides various optimization tools which help to update the parameters based on loss of a model during training.

    # Its include common optimizer such as Adam, RSMprop, Stochastic Gradient Descent (SGD) and more.
    # It handles weight updates effectively and including additional features like learning rate scheduling and Weight Decay(Regularization).

# model.parameters method in PyTorch retrives an iterator all over the trainable parameters (weights and biases) in amodel.
# Parametrs: Weights and biases
# Optimizer use these parameters to claculate gradient and update them during training.
```

```python
# Code  Example:

# Inport Required Libraries
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split
```

```python
# Upload Data set
df=pd.read_csv('breast_cancer.csv')
df.head(4)
```

```python
# Drop the Id column
df=df.drop(columns='id')
df.head(2)
```

```python
# Extract Input and Oputput Value
X=df.iloc[:, 1:]
y=df.iloc[:, 0]
```

```python
# Split the dataset into train and test set
X_train, X_test, y_train, y_test=train_test_split(
    X,y, test_size=0.2
)
```

```python
# Scale the train data
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
# Perform Label Encoding on Target Column
encoder=LabelEncoder()
y_train=encoder.fit_transform(y_train)
y_test=encoder.transform(y_test)
```

```python
# Convert numpy array to PyTorch tensor (float_32)
X_train_tensor = torch.from_numpy(X_train_scaled).float()
X_test_tensor = torch.from_numpy(X_test_scaled).float()

y_train_tensor = torch.from_numpy(y_train).float()
y_test_tensor = torch.from_numpy(y_test).float()
```

```python
# Building the Neural Network using built-in nn.module

class MySimpleNN(nn.Module):

    def __init__(self, num_features):
        super().__init__()

        self.linear = nn.Linear(num_features, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, features):

        out = self.linear(features)
        out = self.sigmoid(out)

        return out
```

```python
# Define Important parameter
learning_rate=0.1
epochs=25
```

```python
# Use Built-in loss function
loss_function=nn.BCELoss()
```

```python
# Create model
model = MySimpleNN(X_train_tensor.shape[1])

# Define Built-in optimizer
optimizer=torch.optim.SGD(model.parameters(), lr=learning_rate)

# Training loop
for epoch in range(epochs):

    # Forward pass
    y_pred = model(X_train_tensor)

    # Calculate loss
    loss = loss_function(y_pred, y_train_tensor.view(-1,1))

    # Clear Gradient
    optimizer.zero_grad()

    # Backward pass
    loss.backward()

    # Parameters update
    optimizer.step()

    # Zero gradients
    model.linear.weight.grad.zero_()
    model.linear.bias.grad.zero_()

    # Print loss
    print(f'Epoch: {epoch + 1}, Loss: {loss.item():.4f}')
```

```python
# model evaluation
with torch.no_grad():
  y_pred = model.forward(X_test_tensor)
  y_pred = (y_pred > 0.9).float()
  accuracy = (y_pred == y_test_tensor).float().mean()
  print(f'Accuracy: {accuracy.item()}')
```
